```python
# Load tasks from file into a list
tasks = []
print("Please upload a text file named 'tasks.txt' to ensure smooth running.")
print("Download it at the end to keep a record of today's tasks in your device! \n")

f = open("tasks.txt", "r")           # open for reading
for line in f:
    task = line.strip()              # remove the new line
    if task:
        tasks.append(task)
f.close()                            # close when done

# Save tasks back to the file
def save_tasks(tasks):
  f = open("tasks.txt", "w")           # open for writing
  for task in tasks:
      f.write(task + "\n")             # write each task + new line
  f.close()                            # close when done

## Defining a function for the menu display
def menu():
  print("Menu:")
  print("1. Add new task.")
  print("2. Show all tasks.")
  print("3. Completed a task.")
  print("4. Remove a task.")
  print("5. Exit.")

## Defining 2nd function to add tasks to list
def add_task(tasks):
  task=input("Enter task: ").strip()
  if task:
    tasks.append(task)
    save_tasks(tasks) ## adds the task to the list
    print("Added!\n")
  else:
    print("Task can not be empty!")

## Defining 3rd function to display tasks
def print_tasks(tasks):
  print("\nYour tasks: ")
  for i, task in enumerate(tasks, 1): ## using enumerate to keep track of tasks
    print(str(i)+".", task) ## printing the status as an icon and tasks in lines

## 4th Function to mark tasks as completed
def complete_task(tasks):
  try: ## using try and except to ensure correct data type entered
    if len(tasks)==0:
      print("Enter some tasks first.")
    else:
      tasknum=int(input("Enter the task number you have complete: "))
      tasknum -=1
      tasks[tasknum]="✓ " + tasks[tasknum]
      save_tasks(tasks)
      print("Congratulations on completing your task!")
      print_tasks(tasks)
  except ValueError:
    print("Error.")

def remove_task(tasks):
  try:
    if len(tasks)==0:
      print("Enter some tasks first.")
    else:
      tasknum=int(input("Enter the task number you want to remove "))
      tasknum -=1
      del tasks[tasknum]
      print("Task deleted.")
      save_tasks(tasks)
      print_tasks(tasks)
  except ValueError:
    print("Error.")

## Interact with user
print("Welcome to your PMU Coursework to-do list!\n")
menu()
selection=0

## Menu selection by user
while selection!=5:
  try:
    selection=int(input("\nEnter the operation you want to procede with (Enter 0 to view the menu again)\n"))
    if selection==0:
      menu()
    elif selection==1:
      ntasks=int(input("How many tasks would you like to add? "))
      for i in range(ntasks):
        add_task(tasks)
    elif selection==2:
      print_tasks(tasks)
    elif selection==3:
      print_tasks(tasks)
      complete_task(tasks)
    elif selection==4:
      print_tasks(tasks)
      ntasks=int(input("How many tasks would you like to remove? "))
      for i in range(ntasks):
        remove_task(tasks)
    elif selection==5:
      print("Thank you for using our program.")
      break
    else:
      print("Try again. Enter the correct number.")
  except ValueError:
    print("Error. Enter the correct number.")
```
